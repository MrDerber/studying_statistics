Изучение matplotlib.pyplot по документации: https://matplotlib.org/stable/tutorials/pyplot.html

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Построение обычного линейного графика с заданными значениями X (значения Y генерятся автоматически):

```python
plt.plot([1, 2, 3, 10])
plt.ylabel('y-axis')
plt.show()
```

Построение аналогичного линейного графика, только теперь с заданными значениями X и Y:

```python
plt.plot([1, 2, 3, 4], [1, 10, 40, 100])
plt.ylabel('y-axis')
plt.xlabel('x-axis')
plt.show()
```

Построение графика с указанием:
1. Типа соединительной линии (прямая линия, пунктирная линия, точки и др.)
2. Цвета

Здесь в документации есть обозначения для соединительных линий, цветов и точек.

```python
plt.title('- - соединить точки линией, o – жирно выделить точки, r – красный цвет (red)')
plt.plot([1, 10, -5, 22], [1, 2, 3, 4], '-or')
plt.axis((-10, +25, 0, +5))#Задание размеров координатной плоскости: (x_min, x_max, y_min, y_max)
plt.show
```

Также можно строить графики на данных на основе объектов, к данным которых обращаешься через текстовое значение.

Например, так обращаются к колонке pandas DataFrame:

```python
df = pd.DataFrame(
    data={
        'a': np.random.randint(low=1, high=101, size=50),
        'b': np.random.randint(low=1, high=101, size=50)
    }
)
df['c'] = df['a']*10
```

```python
df.head()
```

```python
plt.title('x - координата точки по оси X, y - координата точки по оси Y, c - цвет (color) точки, s - размер (size) точки')
plt.scatter(x='a', y='b', c='c', s='c', data=df)
plt.show()
```

Также по оси X могут быть не только числовые, но и категориальные значения:

```python
categories = ['category a', 'category b', 'category c']
values = [1, 5, 20]
```

```python
plt.figure(figsize=(12, 3))

plt.subplot(1, 3, 1)
plt.plot(categories, values)

plt.subplot(1, 3, 2)
plt.bar(categories, values)

plt.subplot(1, 3, 3)
plt.scatter(categories, values)

plt.show()
```

Также можно добавлять надписи прямо на график:

```python
%config InlineBackend.figure_format = 'svg' #Включил векторную графику, чтобы все графики были чётче.
```

```python
mu = 15
sigma = 50

x = mu + sigma * np.random.randn(1000)

plt.hist(x, bins=30)
plt.grid(visible=True)#Добавлена сетка, чтобы было понятнее в каких координатах разместить надпись.
plt.text(x=-170, y=40, s=f'$X \sim N (\mu = {mu}, \sigma = {sigma})$')
plt.axis((-200, 200, 0, 100))
plt.show()
```

Ещё можно делать подписи в виде указателей со стрелками:

```python
x = np.arange(0, 4, 0.01)

y = (x - 2)**4

plt.plot(x, y, color='black')
plt.axis((-2, 6, -4, 16))
plt.grid(True)
plt.annotate(text='минимум функции $f(x) = (x-2)^{2}$', xy=(2.02, -0.14), xytext=(2.5, -2.5), arrowprops=dict(facecolor='green'))
plt.show()
```
